```python
orders = [

    {"id": 2001,
"customer": "Anan",
"city": "Korat",

     "product":
"Keyboard", "qty": 2, "price": 450},

    {"id": 2002,
"customer": "Benja", "city": "Bangkok",

     "product":
"Mouse",    "qty": 5,
"price": 180},

    {"id": 2003,
"customer": "Chai",
"city": "Bangkok",


     "product":
"Monitor",  "qty": 1,
"price": 3200},

    {"id": 2004,
"customer": "Duang", "city": "Phuket",

     "product":
"Mouse",    "qty": 3,
"price": 180},

    {"id": 2005,
"customer": "Anan",
"city": "Korat",

     "product":
"Monitor",  "qty": 2,
"price": 3200},

    {"id": 2006,
"customer": "Ekk",
"city": "Bangkok",

     "product":
"Keyboard", "qty": 4, "price": 450},

]

sales = []
sale_total = 0
for v in orders:
  vat = (v.get('qty') * v.get('price'))*0.07
  total = (v.get('qty') * v.get('price')) + vat
  v["vat"] = vat
  v["total"] = total
  tuple = (v.get('id'), v.get('vat'), v.get('total'))
  sales.append(tuple)
  sale_total += total
for i in sales:
  print(f"ID:{i[0]}")
  print(f"Vat:{i[1]:.2f}")
  print(f"ราคารวม:{i[2]:.2f}")
print(sales)
print(f"ราคารวมทั้งหมด:{sale_total:.2f}")
```
